# FrameChoreo in a notebook
Install FrameChoreo into this kernel from the source checkout or prepared wheel, then run the cells. All data is synthetic.

```python
import pandas as pd

from framechoreo import DataStory

story = DataStory(title="Notebook: where did the total come from?")
story
```

```python
source = story.table(
    pd.DataFrame({"category": ["A", "A", "B"], "amount": [12, 15, 9]}), name="Measurements"
)
total = source.group_sum(by="category", value="amount", dropna=False)
story.annotate(total, note="Select 27 to inspect its two inputs.", hold=4, highlight="amount")
total
```

```python
assert total.to_pandas()["amount"].tolist() == [27, 9]
assert [int(o.value) for o in total.explain(0, "amount")] == [12, 15]
total.to_pandas()
```
